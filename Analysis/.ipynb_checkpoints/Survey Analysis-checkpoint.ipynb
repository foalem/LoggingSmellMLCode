```python
def plot_pie_google_style(df, col, outfile):
    """
    Draws a Google-Forms–style pie chart resembling the attached figure:
    - Two-line labels (category + %)
    - Category in darker text, percentage in light gray
    - Leader lines ending in a gray dot
    - Soft blue gradient wedges
    """

    counts = df[col].dropna().value_counts()
    labels = counts.index
    sizes = counts.values

    # Google-Forms-like soft blue palette
    colors = [
        "#BBD4FF", "#9BBFF9", "#7FA6F9",
        "#5A8FF0", "#3366CC", "#1B4FB3"
    ][:len(sizes)]

    fig, ax = plt.subplots(figsize=(12, 7))

    wedges, _ = ax.pie(
        sizes,
        colors=colors,
        startangle=90,
        wedgeprops={'linewidth': 1.5, 'edgecolor': 'white'},
        normalize=True
    )

    # Loop through each wedge for label placement
    for i, p in enumerate(wedges):

        ang = (p.theta2 - p.theta1) / 2 + p.theta1
        x = np.cos(np.deg2rad(ang))
        y = np.sin(np.deg2rad(ang))

        # Label text styles
        category_text = labels[i]
        percent_text = f"{sizes[i] / sizes.sum() * 100:.1f}%"

        # Position text outside the pie
        x_text = x * 1.25
        y_text = y * 1.25

        # Arrow line pointing to slice
        ax.plot([x * 0.9, x_text], [y * 0.9, y_text], color="gray", lw=1)

        # Small round dot at end of leader line
        ax.scatter([x * 0.9], [y * 0.9], color="gray", s=20)

        # First line: category (darker color, bigger font)
        ax.text(
            x_text, y_text + 0.03,
            category_text,
            fontsize=12,
            ha="left" if x_text > 0 else "right",
            va="center",
            color="black",
        )

        # Second line: percentage (lighter gray)
        ax.text(
            x_text, y_text - 0.03,
            percent_text,
            fontsize=12,
            ha="left" if x_text > 0 else "right",
            va="center",
            color="#999999",
        )

    ax.set_title("")
    ax.set_aspect("equal")

    plt.tight_layout()
    plt.savefig(outfile, dpi=300)
    plt.close()
```

```python
def plot_role_bar_improved(df, outfile):
    col = "What is your current role/title?"
    counts = df[col].dropna().value_counts()
    total = counts.sum()  # total number of valid responses

    fig, ax = plt.subplots(figsize=(8, 3))

    bars = ax.barh(counts.index, counts.values, color="#5A8FF0")

    # Add values + percentages at the end of bars
    for bar in bars:
        width = bar.get_width()
        pct = width / total * 100
        ax.text(
            width + 0.3,
            bar.get_y() + bar.get_height() / 2,
            f"{width} ({pct:.1f}%)",
            va='center',
            fontsize=12
        )

    ax.set_xlabel("Count", fontsize=12)
    ax.invert_yaxis()  # Highest count on top

    plt.tight_layout()
    plt.savefig(outfile, dpi=300)
    plt.close()
```

```python
def plot_multi_select_bar(df, col, outfile):
    """
    Takes a dataframe column containing multiple selections separated by ';'
    and plots a horizontal bar chart with counts and percentages.
    """

    # Drop NaN and split elements
    all_items = (
        df[col]
        .dropna()
        .str.split(";")              # split on semicolon
        .explode()                   # flatten into individual rows
        .str.strip()                 # remove leading/trailing spaces
    )

    counts = all_items.value_counts()
    total = len(df[col].dropna())   # number of respondents

    # Create figure
    fig, ax = plt.subplots(figsize=(12, 2))  # wide and short like example

    bars = ax.barh(counts.index, counts.values, color="#5A8FF0")

    # Add labels with count + percentage
    for bar, label in zip(bars, counts.index):
        width = bar.get_width()
        percent = width / total * 100
        ax.text(
            width + 0.3,
            bar.get_y() + bar.get_height()/2,
            f"{width} ({percent:.1f} %)",
            va='center',
            fontsize=11
        )

    ax.set_xlabel("Count", fontsize=12)
    ax.invert_yaxis()  # largest bar at top

    # Remove top/bottom spines for cleaner look
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    plt.tight_layout()
    plt.savefig(outfile, dpi=300)
    plt.close()
```

```python
# --------------------------------------------------------------
# 1. Canonical List of Logging Smells (normalized for matching)
# --------------------------------------------------------------
SMELL_TOPICS = {
    "Ambiguous logging": "Ambiguous Logging",
    "Misleading Logging": "Misleading Logging",
    "Heavy Logging": "Heavy Logging",
    "Misconfigured Logging": "Misconfigured Logging",
    "Misrouted Metric Logging": "Misrouted Metric Logging",
    "Metric Overwrite": "Metric Overwrite",
    "Log Without Context": "Log Without Context",
    "Missing Hyperparameter Logging": "Missing Hyperparameter Logging",
    "Print-Based Metrics": "Print-Based Metrics",
    "Print Logging": "Print Logging",
    "Logging Sensitive Data": "Logging Sensitive Data",
    "Incorrect Log Level": "Incorrect Log Level",
}

# Normalize keys for matching
NORMALIZED_SMELLS = {k.lower(): v for k, v in SMELL_TOPICS.items()}


# --------------------------------------------------------------
# 2. Mapping rules for the 5 types of survey questions per smell
# --------------------------------------------------------------
QUESTION_PATTERNS = {
    "agree": r"do you agree",                  # Agreement
    "frequency": r"how frequently",            # Frequency
    "relevance": r"how relevant",              # Relevance
    "severity": r"how severe",                 # Severity
    "justification": r"in your own words|explain|describe",  # Open text
}

SHORT_LABEL = {
    "agree": "Agreement",
    "frequency": "Frequency",
    "relevance": "Relevance",
    "severity": "Severity",
    "justification": "Justification",
}


# --------------------------------------------------------------
# 3. Main Header Simplification Function
# --------------------------------------------------------------
def simplify_logging_smell_headers(df: pd.DataFrame) -> pd.DataFrame:
    """
    Simplify Google Forms survey headers for 12 logging smells into compact,
    analysis-ready labels such as:
        - Agreement Ambiguous Logging
        - Frequency Heavy Logging
        - Relevance Metric Overwrite
        - Severity Print Logging
        - Justification Misleading Logging
    """
    new_columns = {}

    for col in df.columns:
        col_clean = clean_text(col)

        smell = detect_smell(col_clean)
        qtype = detect_question_type(col_clean)

        if smell and qtype:
            new_columns[col] = f"{SHORT_LABEL[qtype]} {smell}"
        else:
            # Non-smell-related or demographic question → keep original
            new_columns[col] = col

    return df.rename(columns=new_columns)


# --------------------------------------------------------------
# 4. Detect smell topic referenced in column header
# --------------------------------------------------------------
def detect_smell(text: str) -> str:
    text_lower = text.lower()

    for key, canonical in NORMALIZED_SMELLS.items():
        if key in text_lower:
            return canonical

    return None


# --------------------------------------------------------------
# 5. Detect which type of question (agreement, relevance, etc.)
# --------------------------------------------------------------
def detect_question_type(text: str) -> str:
    text_lower = text.lower()

    for qtype, pattern in QUESTION_PATTERNS.items():
        if re.search(pattern, text_lower):
            return qtype

    return None


# --------------------------------------------------------------
# 6. Pre-clean text for robust matching
# --------------------------------------------------------------
def clean_text(text: str) -> str:
    return " ".join(text.replace("\n", " ").split())
```

```python
import matplotlib.cm as cm
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# NORMALIZATION OF RAW SURVEY RESPONSES → CLEAN LIKERT VALUES
# -------------------------------------------------------------
NORMALIZE_MAP = {
    # Neutral variants
    "Not sure": "Neutral",
    "Not sure / I can’t assess": "Neutral",
    "Not sure / depends heavily on context": "Neutral",

    # Frequency variants
    "Rarely (a few times per year)": "Rarely",
    "Sometimes (a few times per quarter)": "Sometimes",
    "Often (monthly)": "Often",
    "Very often (weekly or more)": "Very often",

    # Relevance variants
    "Not relevant at all": "Not relevant",
    "Minor": "Slightly severe",
    "Moderate": "Moderately severe",
    "Severe": "Very severe",
    "Critical": "Extremely severe"
}

def normalize_likert_columns(df):
    df = df.copy()
    for col in df.columns:
        # Only normalize smell-related Likert columns
        if col.startswith(("Agreement", "Frequency", "Relevance", "Severity")):
            df[col] = df[col].replace(NORMALIZE_MAP)
    return df

# df = normalize_likert_columns(df)

LIKERT_ORDER = [
    "Strongly disagree", "Neutral", "Disagree", "Agree", "Strongly agree",
    "Never", "Rarely", "Sometimes", "Often", "Very often", "Always",
    "Not relevant", "Slightly relevant", "Moderately relevant", 
    "Very relevant", "Extremely relevant",
    "Not severe", "Slightly severe", "Moderately severe",
    "Very severe", "Extremely severe"
]

def detect_likert_options(df, metric_prefix):
    cols = [c for c in df.columns if c.startswith(metric_prefix)]
    values = pd.unique(df[cols].values.ravel())
    values = [v for v in values if isinstance(v, str)]

    # Filter Likert ordering for only values found in data
    ordered = [opt for opt in LIKERT_ORDER if opt in values]
    return ordered

def build_metric_matrix(df, metric, options):
    """
    Build matrix for stacked 100% bar chart.
    Rows = smells
    Columns = Likert options
    """

    # Get all smell names from column pattern
    smell_columns = [
        col for col in df.columns
        if metric in col and "Severity" not in col and "Frequency" not in col
    ]

    topic_names = []
    matrix = []

    for col in smell_columns:
        topic = col.split(metric)[0].strip()
        topic_names.append(topic)

        counts = []
        total = df[col].notna().sum()

        for opt in options:
            count = (df[col] == opt).sum()
            percentage = (count / total) * 100 if total > 0 else 0
            counts.append(percentage)

        matrix.append(counts)

    return topic_names, np.array(matrix)

def make_colors(n):
    cmap = plt.get_cmap("viridis")
    return [cmap(i / (n - 1)) for i in range(n)]


def plot_stacked_100pct(topic_names, matrix, options, outfile):

    n_topics = len(topic_names)
    percentages = matrix / matrix.sum(axis=1, keepdims=True) * 100

    fig, ax = plt.subplots(figsize=(12, 6))
    colors = make_colors(len(options))
    left = np.zeros(n_topics)

    for i, opt in enumerate(options):
        segment = percentages[:, i]
        ax.barh(topic_names, segment, left=left,
                color=colors[i], label=opt)

        # Counts inside segments
        for y in range(n_topics):
            count = matrix[y, i]
            pct = segment[y]
            if count > 0 and pct > 3:
                ax.text(left[y] + pct/2, y, f"{count}",
                        ha="center", va="center", fontsize=9)

        left += segment

    ax.set_xlabel("Percentage (%)")
    ax.set_xlim(0, 100)

    # ------------------------------
    # LEGEND ON TOP (horizontal)
    # ------------------------------
    ax.legend(
        ncol=len(options),
        loc="upper center",
        bbox_to_anchor=(0.5, 1.20),
        frameon=False,
        fontsize=10
    )

    # Remove title
    ax.set_title("")

    plt.tight_layout()
    plt.savefig(outfile, dpi=300, bbox_inches="tight")
    plt.close()
    
    
    
df = normalize_likert_columns(df_clean)

for metric in ["Agreement", "Frequency", "Relevance", "Severity"]:
    options = detect_likert_options(df, metric)
    topic_names, matrix = build_metric_matrix(df, metric, options)

    plot_stacked_100pct(
        topic_names,
        matrix,
        options,
        outfile=f"{metric.lower()}_stacked_bar.pdf"
    )




# df = df_clean
# # ---------------------------------------------------------
# # AUTO-DETECT LIKERT OPTIONS
# # ---------------------------------------------------------
# def detect_likert_options(df, metric_prefix):
#     """Automatically infer all unique Likert responses for this metric."""
#     cols = [c for c in df.columns if c.startswith(metric_prefix)]
#     values = pd.unique(df[cols].values.ravel())
#     values = [v for v in values if isinstance(v, str)]
#     return sorted(values)

# # ---------------------------------------------------------
# # COLOR PALETTE
# # ---------------------------------------------------------
# def make_colors(n):
#     cmap = cm.get_cmap("Blues", n + 2)
#     return [cmap(i + 1) for i in range(n)]


# # ---------------------------------------------------------
# # STACKED 100% BAR PLOT
# # ---------------------------------------------------------
# def plot_stacked_100pct(topic_names, matrix, options, outfile):

#     n_topics = len(topic_names)
#     percentages = matrix / matrix.sum(axis=1, keepdims=True) * 100

#     fig, ax = plt.subplots(figsize=(12, 7))
#     colors = make_colors(len(options))
#     left = np.zeros(n_topics)

#     for i, opt in enumerate(options):
#         segment = percentages[:, i]
#         ax.barh(
#             topic_names,
#             segment,
#             left=left,
#             color=colors[i],
#             label=opt
#         )

#         # Add count labels inside bars
#         for y in range(n_topics):
#             count = matrix[y, i]
#             pct = segment[y]
#             if count > 0 and pct > 3:  
#                 ax.text(left[y] + pct / 2, y, f"{count}",
#                         ha="center", va="center", fontsize=9)

#         left += segment

#     # Axis label
#     ax.set_xlabel("Percentage (%)")
#     ax.set_xlim(0, 100)

#     # ---------------------------------------------------------
#     # LEGEND ON TOP (centered, horizontal)
#     # ---------------------------------------------------------
#     ax.legend(
#         ncol=len(options),               # spread horizontally
#         loc="upper center",
#         bbox_to_anchor=(0.5, 1.15),     # place above plot
#         frameon=False,
#         fontsize=10
#     )

#     # Remove title
#     ax.set_title("")

#     plt.tight_layout()
#     plt.savefig(outfile, dpi=300, bbox_inches="tight")
#     plt.close()


# # ---------------------------------------------------------
# # GENERATE ALL FIGURES
# # ---------------------------------------------------------
# for metric in ["Agreement", "Frequency", "Relevance", "Severity"]:
#     options = detect_likert_options(df, metric)
#     topic_names, matrix = build_metric_matrix(df, metric, options)

#     plot_stacked_100pct(
#         topic_names,
#         matrix,
#         options,
#         outfile=f"{metric.lower()}_stacked_bar.pdf"
#     )
```
